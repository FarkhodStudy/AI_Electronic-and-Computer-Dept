```python
def step(z):
    if z > 0:
        y = 1
    else:
        y = 0
    return y
def NOT(x1):
    z = -x1 + 0.5          # Weight w = -1, Bias b = 0.5
    return step(z)
print("=== NOT Gate Truth Table ===")
print(f"NOT(0) = {NOT(0)}") # Input 0 -> Output 1
print(f"NOT(1) = {NOT(1)}") # Input 1 -> Output 0
```

```python
def step(z):
    if z > 0:
        y = 1
    else:
        y = 0
    return y
def AND(x1, x2):
    z = x1 + x2 - 1.5    # Weights w1 = 1, w2 = 1, Bias b = -1.5
    return step(z)
print("=== AND Gate Truth Table ===")
print(f"AND(0, 0) = {AND(0, 0)}")  # Input (0, 0) -> Output 0
print(f"AND(0, 1) = {AND(0, 1)}")  # Input (0, 1) -> Output 0
print(f"AND(1, 0) = {AND(1, 0)}")  # Input (1, 0) -> Output 0
print(f"AND(1, 1) = {AND(1, 1)}")  # Input (1, 1) -> Output 1
```

```python
def step(z):
    if z > 0:
        y = 1
    else:
        y = 0
    return y

def OR(x1, x2):
    z = x1 + x2 - 0.5    # Weights w1 = 1, w2 = 1, Bias b = -0.5
    return step(z)

print("=== OR Gate Truth Table ===")
print(f"OR(0, 0) = {OR(0, 0)}")  # Input (0, 0) -> Output 0
print(f"OR(0, 1) = {OR(0, 1)}")  # Input (0, 1) -> Output 1
print(f"OR(1, 0) = {OR(1, 0)}")  # Input (1, 0) -> Output 1
print(f"OR(1, 1) = {OR(1, 1)}")  # Input (1, 1) -> Output 1
```

```python
def step(z):
    if z > 0:
        y = 1
    else:
        y = 0
    return y

def NOT(x1):
    z = -x1 + 0.5
    return step(z)

def AND(x1, x2):
    z = x1 + x2 - 1.5
    return step(z)

def OR(x1, x2):
    z = x1 + x2 - 0.5
    return step(z)

def XOR(x1, x2):
    # XOR(x1, x2) = (x1 OR x2) AND NOT(x1 AND x2)
    h1 = OR(x1, x2)
    h2 = NOT(AND(x1, x2))
    return AND(h1, h2)

print("=== XOR Gate Truth Table ===")
print(f"XOR(0, 0) = {XOR(0, 0)}")  # Input (0, 0) -> Output 0
print(f"XOR(0, 1) = {XOR(0, 1)}")  # Input (0, 1) -> Output 1
print(f"XOR(1, 0) = {XOR(1, 0)}")  # Input (1, 0) -> Output 1
print(f"XOR(1, 1) = {XOR(1, 1)}")  # Input (1, 1) -> Output 0
```

```python
import numpy as np
from sklearn.linear_model import Perceptron

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])          # XOR targets

perc = Perceptron(random_state=42)  # one neuron
perc.fit(X, y)
print("Predictions:", perc.predict(X))
print(f"Accuracy: {perc.score(X, y):.0%}")
```

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(
    hidden_layer_sizes=(2,),   # h1 and h2
    activation="logistic",     # smooth step
    solver="lbfgs",            # good for tiny data
    random_state=0,            # same start for all
)
```

```python
mlp.fit(X, y)      # training happens here

print("Predictions:", mlp.predict(X))
print(f"Accuracy: {mlp.score(X, y):.0%}")

proba = mlp.predict_proba(X)[:, 1]   # P(y = 1)
print("P(y=1):", proba.round(3))
```

```python
W1, W2 = mlp.coefs_        # weights (2 layers)
b1, b2 = mlp.intercepts_   # biases  (2 layers)
print("W1 =\n", W1.round(1))
print("b1 =", b1.round(1))
print("W2 =", W2.ravel().round(1))
print("b2 =", b2.round(1))
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

H = sigmoid(X @ W1 + b1)   # hidden layer
print(H.round(2))
```
