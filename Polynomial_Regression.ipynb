```python
#######
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract the coefficients in the form f(x) = ax^2 + bx + c
a, b, c = mymodel.coefficients

# Define the prediction function
def f(x_value):
    return mymodel(x_value)

# Print the prediction equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Generate x-values for prediction
myline = np.linspace(1, 22, 100)

# Visualization
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data from previous cell
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Predict the speed at 17:00 (x = 17)
target_time = 17
predicted_speed = mymodel(target_time)

# 1. Print the predicted speed
print(f"Predicted speed at 17:00 (5:00 PM): {predicted_speed:.2f}")

# Generate x-values for plotting the line
myline = np.linspace(1, 22, 100)

# 2. Visualization: Display the predicted value on the graph
plt.scatter(x, y, label='Actual Data')
plt.plot(myline, mymodel(myline), color='red', label='Polynomial Fit')
plt.scatter(target_time, predicted_speed, color='green', s=150, zorder=5, label=f'Prediction at 17:00 ({predicted_speed:.1f})')

plt.title("Polynomial Regression with Prediction")
plt.xlabel("Time Farkhod")
plt.ylabel("Speed")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
